Introduction to Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3]
y = [5,10,20]

plt.plot(x, y, label = 'Showers', color = 'green', marker = '^', markersize = 15, linestyle = '--', markeredgecolor = 'blue')

plt.title('Matplotlib Viz', fontdict = {'fontname': 'Times New Roman', 'fontsize': 30})
plt.xlabel('X Axis')
plt.ylabel('Y Axis')


plt.xticks([1,2,3])
plt.yticks([0,5,10,15,20])
plt.legend()

plt.show()
```

```python
x = [1,2,3]
y = [5,10,20]

plt.plot(x, y, label = 'Showers', color = 'green', marker = '^', markersize = 15, linestyle = '--', markeredgecolor = 'blue')

z = [1,5,10]
plt.plot(x, z, label = 'Bath', color = 'red', marker = '.', markersize = 20, linestyle = '--', markeredgecolor = 'green')

plt.title('Matplotlib Viz', fontdict = {'fontname': 'Times New Roman', 'fontsize': 30})
plt.xlabel('X Axis')
plt.ylabel('Y Axis')


plt.xticks([1,2,3])
plt.yticks([0,5,10,15,20])
plt.legend()

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
smoking = pd.read_csv(r'C:\Users\khairil.akub\Desktop\Mohd Khairil\Personal\Personal Development\Python\Pandas\Data Visualization\smoking.csv',
                     encoding = 'ISO-8859-1')

smoking
```

```python
smoker = smoking[smoking['smoke'] == 'Yes']
```

```python
female_smokers = smoker[smoker['gender'] == 'Female']
male_smokers = smoker[smoker['gender'] == 'Male']
```

```python
fig, axs = plt.subplots(1,2)

sns.histplot(female_smokers['age'], bins = 25, ax = axs[0])
axs[0].set_title('Age Distribution of Female Smokers')

sns.histplot(male_smokers['age'], bins = 25, ax = axs[1])
axs[1].set_title('Age Distribution of Male Smokers')

plt.tight_layout()
plt.show()
```

```python
smoking.corr(numeric_only = True)
```

```python
housing = pd.read_csv(r'C:\Users\khairil.akub\Desktop\Mohd Khairil\Personal\Personal Development\Python\Pandas\Data Visualization\real_estate.csv',
                     encoding = 'ISO-8859-1')
```

```python
housing_filt = housing.drop('zip_code', axis = 1)
```

```python
correlation = housing_filt.corr(numeric_only = True)
```

```python
sns.heatmap(correlation, annot = True, cmap = 'coolwarm')
```

```python
correlation
```

```python
#normaldistribution
pearson_correlation = housing_filt.corr(numeric_only = True, method = 'pearson')
pearson_correlation
```

```python
#nottakingoutlier
spearman_correlation = housing_filt.corr(numeric_only = True, method = 'spearman')
spearman_correlation
```

```python
#strictmoreonoutlier
kendall_correlation = housing_filt.corr(numeric_only = True, method = 'kendall')
kendall_correlation
```

```python
sns.heatmap(pearson_correlation, annot = True, cmap = 'coolwarm')
plt.title('Pearson Correlation')
plt.show()
```

```python
sns.heatmap(spearman_correlation, annot = True, cmap = 'coolwarm')
plt.title('Spearman Correlation')
plt.show()
```

```python
sns.heatmap(kendall_correlation, annot = True, cmap = 'coolwarm')
plt.title('Kendall Correlation')
plt.show()
```

```python
fig, axs = plt.subplots(3,1, figsize = (5,15))

sns.heatmap(pearson_correlation, annot = True, cmap = 'coolwarm', ax = axs[0])
axs[0].set_title('Pearson Correlation')

sns.heatmap(spearman_correlation, annot = True, cmap = 'coolwarm', ax = axs[1])
axs[1].set_title('Spearman Correlation')

sns.heatmap(kendall_correlation, annot = True, cmap = 'coolwarm', ax = axs[2])
axs[2].set_title('Kendall Correlation')

plt.tight_layout()
plt.show()
```
