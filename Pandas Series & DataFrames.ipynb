Pandas for Dataframe

```python
import pandas as pd

index_value = ["Best", "Second Best", "Third Best"]
my_list = [10,25,35]

one = pd.Series(data = my_list, index = index_value)
one
```

```python
index_value2 = ["Best", "Second Best", "Forth Best"]
my_list2 = [10,25,40]

two = pd.Series(data = my_list2, index = index_value2)
two
```

```python
dataframe = pd.DataFrame({'Series 1': one, 'Series 2': two})
```

```python
print(dataframe)
```

```python
dataframe
```
